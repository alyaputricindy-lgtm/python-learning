```python
# Contoh Sederhana

nilai = int(input("Masukkan nilai Anda:"))

if nilai >= 90:
   print("Nilai A")
elif nilai >= 85:
   print("Nilai B+")
elif nilai >= 80:
   print("Nilai B")
elif nilai >= 75:
   print("Nilai C+")
elif nilai >= 70:
   print("Nilai C")
else:
   print("Nilai Anda Tidak Lulus")
   print("Silahkan mengajukan remidi")
```

```python
# Latihan

nama_mahasiswa = input("Nama mahasiswa: ")
nilai_tugas = int(input("Masukkan nilai tugas: "))
nilai_uts = int(input("Masukkan nilai UTS: "))
nilai_uas = int(input("Masukkan nilai UAS:: "))

nilai_akhir = (nilai_tugas * 20/100) + (nilai_uts * 30/100) + (nilai_uas * 50/100)
print(f"Nilai akhir: {nilai_akhir}")

if nilai_akhir >= 85:
   print("Grade: A")
elif nilai_akhir >= 70:
   print("Grade: B")
elif nilai_akhir >= 60:
   print("Grade: C")
else:
   print("Grade: D")
```

```python
# Latihan

vip = 500000
reguler = 250000
ekonomi = 100000

jenis_tiket = (input("Masukkan jenis tiket (VIP, Reguler, Ekonomi): "))
jumlah_tiket = int(input("Masukkan jumlah tiket yang dibeli: "))

if jenis_tiket == "VIP":
   total_bayar = vip * jumlah_tiket
elif jenis_tiket == "Reguler":
   total_bayar = reguler * jumlah_tiket
elif jenis_tiket == "Ekonomi":
   total_bayar = ekonomi * jumlah_tiket
else:
   total_bayar = 0
   print("Jenis tiket tidak tersedia")

print(f"Total bayar: {total_bayar}")
```

```python
# Latihan

berat_badan = float(input("Masukkan berat badan (kg): "))
tinggi_badan = float(input("Masukkan tinggi badan (m): "))

BMI = berat_badan * (tinggi_badan ** 2)
print(f"BMI Anda adalah {BMI}")

if BMI < 18.5:
   print(f"Kategori BMI Anda adalah Kurus")
elif BMI < 25:
   print(f"Kategori BMI Anda adalah Normal")
elif BMI < 30:
   print(f"Kategori BMI Anda adalah Gendut")
else:
   print(f"Kategori BMI Anda adalah Obesitas")
```
