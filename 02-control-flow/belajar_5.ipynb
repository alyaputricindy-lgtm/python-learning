# For

Gunakan for jika jumlah perulangan sudah diketahui.

Perbandingannya: For jumlahnya 5, While itu ya 5

```python
# Contoh Sederhana
# Perbandingan 1

for i in range(5):
    print("Halo")
```

```python
# Contoh Sederhana
# Perbandingan 2

for i in range(5):
    print (i)
# Dimulai dari 0
# Jumlahnya ada 5
```

```python
# Contoh Sederhana
# Perbandingan 3

for i in range (1, 6):
    print(i)

# dimulai dari 1
# ditutup sebelum angka 6, yaitu 5
```

```python
# Contoh Sederhana
# Perbandingan 4
# Menampilkan angka genap

for i in range (2, 10, 2):
    print(i)

# dimulai dari 2
# ditutup sebelum angka 10
# kelipatan 2 saja yang ditampilkan
```

```python
# Contoh Sederhana
# Perbandingan 5
# Menampilkan angka ganjil

for i in range (1, 10, 2):
    print(i)

# dimulai dari 1
# ditutup sebelum angka 10
# kelipatan 2 karena dimulai dari 1 sehingga hanya angka ganjil
```

```python
# Contoh Sederhana
# Perbandingan 6
# Menampilkan Hitungan ke 3

for i in range (1, 4):
    print("Hitungan ke-", i)
```

```python
# Contoh Sederhana
# Perbandingan 7
# Hitungan Mundur

for i in range (5, 0, -1):
    print(i)
```

```python
# Contoh Sederhana
# Membuat Segitiga dari *

for i in range(1,6):
    print("*" * i)
```

```python
# Contoh Sederhana
# Membuat Segitiga Terbalik dari *

for i in range (5,0,-1):
    print("*" * i)
```

```python
# Contoh Sederhana
# Membuat Segitiga dari * Menggunakan Nested
# Nested bertujuan agar bisa membuat pola yang sama dengan tambahan berbeda

for i in range(1, 6): # Mengatur jumlah baris
    for j in range(i): # Mengatur jumlah bintang pada setiap baris
        print("*", end="") # Membuat bintang dicetak di baris yang sama.
    print() # yang kosong digunakan untuk pindah ke baris berikutnya.
```

```python
# Contoh Sederhana
# Membuat Segitiga dari * Menggunakan Nested
# Disini meminta spasi antar *

for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Contoh Sederhana
# Segitiga Angka

for i in range(1,6):
    print(str(i) * i)

# range() bisa langsung menghasilkan angka
# langsung diubah menjadi tipe string
# sehingga * fungsinya tidak lagi menjadi perkalian tapi perulangan
```

```python
# Contoh Sederhana
# Segitiga Huruf

huruf = ["A", "B", "C", "D", "E"]
for i in range(5):
    print(huruf[i] * (i+1))
# range() tidak bisa menghasilkan huruf sehingga perlu huruf[i]

"""
i=0
huruf[0] * (i+1) =
A * (0+1) =
A * 1 = A
begitu seterusnya
"""

"""
kenapa i+1?
karena range(5) itu menghasilkan 0 1 2 3 4
nanti jadinya di putaran pertama A * 0
kalau i+1 menjadi A * (0+1) = A * 1
"""
```

```python
# Contoh Sederhana
# Membuat Kotak Bintang

for i in range(5):
    print("*" * 5)
```

```python
# Contoh Sederhana
# Jumlah Hari

hari = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat"]

jumlah = 0

for h in hari:
    jumlah += 1

print("Jumlah hari:", jumlah)
```

# While

```python
# Contoh Sederhana
# Perbandingan 1

i = 0
while i < 5:
    print("Halo")
    i += 1
```

```python
# Contoh Sederhana
# Perbandingan 2

i = 0
while i < 5:
    print (i)
    i += 1
# i dimulai dari 0
# i tidak sampai 5 sehingga hanya sampai 4
```

```python
# Contoh Sederhana
# Perbandingan 2.2

i = 1
while i < 5:
    print (i)
    i += 1
# i dimulai dari 1
# i tidak sampai 5 sehingga hanya sampai 4
```

```python
# Contoh Sederhana
# Perbandingan 2.3

i = 0
while i <= 5:
    print (i)
    i += 1
# i dimulai dari 0
# i sampai dengan 5
```

```python
# Contoh Sederhana
# Perbandingan 2.4

i = 1
while i <= 5:
    print (i)
    i += 1
# i dimulai dari 1
# i sampai dengan 5
```

```python
# Contoh Sederhana
# Perbandingan 3

i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
# Contoh Sederhana
# Perbandingan 4
# Menampilkan angka genap

i = 2
while i < 10:
    print(i)
    i += 2

# dimulai dari 2
# kurang dari 10
# kelipatan 2
```

```python
# Contoh Sederhana
# Perbandingan 5
# Menampilkan angka ganjil

i = 1
while i < 10:
    print(i)
    i += 2

# dimulai dari 1
# kurang dari 10
# kelipatan 2 karena dimulai dari 1 shg hanya angka ganjil
```

```python
# Contoh Sederhana
# Perbandingan 6
# Menampilkan Hitungan ke-3

i = 1
while i <= 3:
    print("Hitungan ke-", i)
    i += 1
```

```python
# Contoh Sederhana
# Perbandingan 7
# Hitungan Mundur

i = 5
while i > 0:
    print(i)
    i -= 1
```

```python
# Projek1
# Menghitung total nilai siswa
# Guru ingin memasukkan nilai 5 siswa
# Cara Manual

# Ketentuan:
# 1. Gunakan for.
# 2. Input nilai sebanyak 5 kali.
# 3. Hitung total nilai.
# 4. Hitung rata-rata.


siswa1 = int(input("Masukkan nilai siswa 1: "))
siswa2 = int(input("Masukkan nilai siswa 2: "))
siswa3 = int(input("Masukkan nilai siswa 3: "))
siswa4 = int(input("Masukkan nilai siswa 4: "))
siswa5 = int(input("Masukkan nilai siswa 5: "))

total = siswa1 + siswa2 + siswa3 + siswa4 + siswa5
print(f"Total nilai: {total}")

ratarata = total / 5
print(f"Rata-rata: {ratarata}")
```

```python
# Projek1
# Cara For

total = 0 # dimulai dari 0
for i in range(5): # rangenya 5 karena jumlah siswa adalah 5
    nilai = int(input(f"Masukkan nilai siswa {i+1}: ")) # biar i nambah 1
    total += nilai
    # total = total + nilai
    # nanti total = 0 + 80
    # terus loop 80 + 90
    # 90 + 75 dan seterusnya

ratarata = total / 5

print(f"Total nilai: {total}")
print(f"Rata-rata: {ratarata}")
```

```python
# Projek 2
# Kasir Mini
# Pembeli Membeli 3 barang
# Ini caraku

# Ketentuan:
# 1. Gunakan for.
# 2. Input harga barang sebanyak 3 kali.
# 3. Hitung total.
# 4. Jika total ≥ Rp100.000 → Diskon 5%.
# 5. Jika total ≥ Rp300.000 → Diskon 10%.
# 6. Selain itu → Tidak ada diskon.

total = 0
for i in range(3):
    harga_barang = int(input("Masukkan harga barang: "))
    total += harga_barang

if total >= 300000:
   diskon = "10%"
   setelah_diskon = int(total * 10/100) # agar hasilnya tidak float1
elif total >= 100000:
   diskon = "5%"
   setelah_diskon = int(total * 5/100) # agar hasilnya tidak float
else:
   diskon = "0%"
   setelah_diskon = 0

print()
print("=" * 25)
print(f"Total: Rp {total}")
print(f"Diskon: {diskon}")
print(f"Bayar: Rp {total - setelah_diskon}")
print("=" * 25)
```

```python
# Projek 2
# Kasir Mini
# Pembeli Membeli 3 barang
# Ini caranya Kenzo

total = 0
for i in range(3):
    harga_barang = int(input("Masukkan harga barang: "))
    total += harga_barang

if total >= 300000:
   diskon = 10 # dengan cara ini diskon tidak bernilai string tapi int
elif total >= 100000:
   diskon = 5
else:
   diskon = 0

setelah_diskon = int(total * diskon/100) # lebih praktis diluar if elif

print()
print("=" * 25)
print(f"Total: Rp {total}")
print(f"Diskon: {diskon}%") # persennya ditaruh disini
print(f"Bayar: Rp {total - setelah_diskon}")
print("=" * 25)
```

```python
# Projek 3
# Login sederhana
# Program akan meminta password sampai benar.
# Pakai while ya berarti

password = ""
# jangan pakai password = 0 karena itu untuk int sedangkan ini ada string
# sehingga password = "" yakni pakai string kosongan

while password != "python123":
    password = input("Masukkan password: ")
# tidak pakai += 1 karena tidak sedang menghitung angka

print("Login berhasil")
```

```python
# Projek 4
# Tebak angka
# Ini caraku

# Caranya adalah
# Buat program:
# Masukkan angka:

# Jika bukan angka 7
# Salah.
# Program akan terus berjalan.

# Kalau angka 7
# Selamat!
# Program selesai.

# Menggunakan while ya

angka = 0
while angka != 7:
    angka = int(input("Masukkan angka: "))
    if angka != 7:
      print("Salah")
    else:
      print("Selamat!")
```

```python
# Projek 4
# Ini cara Kenzo

angka = 0
# pakai angka = 0 karena berupa int/angka

while angka != 7:
    angka = int(input("Masukkan angka: "))

    if angka == 7:
        print("Selamat!") # langsung mengecek kondisi yg diinginkan jd LEBIH SIMPEL
    else:
        print("Salah")
```
