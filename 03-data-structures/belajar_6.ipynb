# Mengenal Struktur Data List

Cara menyimpan banyak data dalam satu variabel adalah dengan menggunakan LIST.

```python
# Cara membuat list di Python
# 1. Membuat list kosong
warna =[]

# 2. Membuat list dengan mengisi 1 item
warna = ["Merah"]

# 3. Membuat list dengan mengisi lebih dari 1 item
warna = ["Merah", "Pink", "Biru"]
```

```python
# Jenis tipe data pada list

laci = ["buku", 21, True, 34.12]

"""
1. Tipe data string
2. Tipe data integer
3. Tipe data boolean
4. Tipe data float
"""
```

```python
# Cara mengambil nilai dari list
buah = ["Apel", "Anggur", "Melon"]

print(buah[2])

"""
[0] adalah indeks ke-0 = Apel
[1] adalah indeks ke-1 = Anggur
[2] adalah indeks ke-2 = Melon
"""
```

```python
# Mengubah nilai list
# List sifatmya mutable (bisa berubah-ubah)
buah = ["Apel", "Anggur", "Melon"]

buah[2] = "Jeruk"
# indeks kedua yaitu ketiga yang Melon diganti jadi Jeruk

print(buah)
```

```python
# Menambahkan item list (1)
# Menambahkan item di belakang
buah = ["Apel", "Anggur", "Melon"]

buah.append("Jeruk")

print(buah)
```

```python
# Menambahkan item list (2)
# Menambahkan item di depan

buah = ["Apel", "Anggur", "Melon"]
buah.insert(0, "Anggur")
# 0 itu indeks ke 0. Berarti disisipkan menjadi indeks ke 0

print(buah)
```

```python
# Menambahkan item di list (3)
# Menambahkan item di indeks tertentu

buah = ["Apel", "Anggur", "Melon"]
buah.insert(2, "Jeruk")
# 2 itu indeks ke 2. Berarti disisipkan menjadi indeks ke 2

print(buah)
```

```python
# Menghapus item di list
# Menghapus berdasarkan kata
buah = ["Apel", "Anggur", "Melon"]
buah.remove("Anggur") # Menghapus item Anggur

print(buah)
```

```python
# Menghapus item di list
# Menghapus berdasarkan index
buah = ["Apel", "Anggur", "Melon"]
buah.pop(1) # Menghapus index ke 1, yaitu Anggur

print(buah)
```

```python
# Menghapus beberapa item di list sekaligus
buah = ["Apel", "Anggur", "Melon"]
del buah[0:2]
# Menghapus index 0 dan sebelum indeks 2, yaitu indeks 0 dan indeks 1

print(buah)
```

```python
# Menghapus semua item
buah = ["Apel", "Anggur", "Melon"]
buah.clear()

print(buah)
```

```python
# Memotong item list
buah = ["Apel", "Anggur", "Melon"]
buah_baru = buah[0:2]
# Mengambil indeks ke 0 sampai sebelum indeks ke 2 saja, yaitu indeks 0 dan indeks 1

print(buah_baru)
```

```python
# Projek
# To do list sederhana
"""
Terdapat fitur:
1. Menambah tugas
2. Melihat semua tugas
3. Menghapus tugas
4. Keluar dari program
"""

"""
===== TO DO LIST =====
1. Tambah Tugas
2. Lihat Tugas
3. Hapus Tugas
4. Keluar

Pilih menu : 1

Masukkan tugas :
Belajar Python

Tugas berhasil ditambahkan!
"""

tugas = []

while True:
  print("===== To Do List =====")
  print("1. Tambah Tugas")
  print("2. Lihat Tugas")
  print("3. Hapus Tugas")
  print("4. Keluar")

  pilihan = int(input("Pilih menu: "))

  if pilihan == 1:
    tugas_baru = input("Tugas baru: ")
    tugas.append(tugas_baru)
    print("Tugas berhasil ditambahkan")

  elif pilihan == 2:
    print("\n===== Daftar Tugas =====")
    if len(tugas) == 0:
      print("Belum ada tugas")
    else:
      for i in range(len(tugas)):
        print(f"{i + 1}. {tugas[i]}")

  elif pilihan == 3:
    print("\n===== Hapus Tugas =====")
    if len(tugas) == 0:
      print("Tidak ada tugas yang bisa dihapus")
    else:
      for i in range(len(tugas)):
        print(f"{i + 1}. {tugas[i]}")

      nomor = int(input("Hapus nomor: "))
      tugas.pop(nomor -1)
      print("Tugas berhasil dihapus")

  elif pilihan == 4:
    print("Terima kasih")
    break

  else:
    print("Pilih sing nggenah woy")
```

```python
# Projek
# Daftar belanja

"""
Fitur =
===== DAFTAR BELANJA =====
1. Tambah Barang
2. Lihat Barang
3. Hapus Barang
4. Jumlah Barang
5. Keluar

Contoh =
Pilih menu : 1
Masukkan barang :
Minyak
Barang berhasil ditambahkan.
"""

troli = []

while True:
  print("===== DAFTAR BELANJA =====")
  print("1. Tambah Barang")
  print("2. Lihat Barang")
  print("3. Hapus Barang")
  print("4. Jumlah Barang")
  print("5. Keluar")

  pilihan = int(input("Pilih menu: "))

  if pilihan == 1:
    barang_baru = input("Masukkan barang: ")
    troli.append(barang_baru)
    print("Barang berhasil ditambahkan")

  elif pilihan == 2:
    print("\n===== Daftar Belanja =====")
    if len(troli) == 0:
      print("Belum ada barang yang ditambahkan")
    else:
      for i in range(len(troli)):
        print(f"{i+1}. {troli[i]}")

  elif pilihan == 3:
    print("\n===== Hapus Barang =====")
    if len(troli) == 0:
      print("Barang tidak ada yang bisa dihapus")
    else:
      for i in range(len(troli)):
        print(f"{i+1}. {troli[i]}")
      nomor = int(input("Hapus nomor:"))
      troli.pop(nomor -1)
      print("Barang berhasil dihapus")

  elif pilihan == 4:
    print("\n===== Jumlah Barang =====")
    if len(troli) == 0:
      print("Jumlah barang masih 0")
    else:
      print(f"Saat ini jumlah barang adalah {len(troli)}")

  elif pilihan == 5:
    print("Terima kasih, Zidan")
    break

  else:
    print("Pilihanmu ra ono, sing nggenah woi")
```

```python
# Projek
# Perpustakaan Mini

"""
Menu=

===== PERPUSTAKAAN =====

1. Tambah Buku
2. Lihat Buku
3. Hapus Buku
4. Cari Jumlah Buku
5. Keluar
"""

rak = []

while True:
  print("===== PERPUSTAKAAN MINI =====")
  print("1. Tambah Buku")
  print("2. Lihat Buku")
  print("3. Hapus Buku")
  print("4. Cari Jumlah Buku")
  print("5. Keluar")

  pilihan = int(input("Pilih menu: "))

  if pilihan == 1:
    buku = input("Masukkan nama buku:")
    rak.append(buku)
    print("Buku berhasil ditambahkan")

  elif pilihan == 2:
    print("\n===== Daftar Buku =====")
    if len(rak) == 0:
      print("Belum ada buku yang ditambahkan")
    else:
      for i in range(len(rak)):
        print(f"{i+1}. {rak[i]}")

  elif pilihan == 3:
    print("\n===== Hapus Buku =====")
    if len(rak) == 0:
      print("Tidak ada buku yang bisa dihapus")
    else:
      for i in range(len(rak)):
        print(f"{i+1}. {rak[i]}")
      hapus = int(input("Hapus nomor:"))
      rak.pop(hapus -1)
      print("Buku berhasil dihapus")

  elif pilihan == 4:
    print(f"Jumlah buku ada {len(rak)}")

  else:
    print("Selesai!")
    break
```

```python
# Projek
# Jadwal Mingguan

jadwal =[]

while True:
  print("===== JADWAL =====")
  print("1. Tambah Jadwal")
  print("2. Lihat Jadwal")
  print("3. Sisipkan Jadwal")
  print("4. Hapus Jadwal")
  print("5. Tampilkan 3 Jadwal Pertama")
  print("6. Keluar")

  pilihan = int(input("Pilih menu: "))

  if pilihan == 1:
    jadwal_tambahan = input("Masukkan jadwal: ")
    jadwal.append(jadwal_tambahan)
    print("Jadwal berhasil ditambahkan")

  elif pilihan == 2:
    print("\n===== Daftar Jadwal =====")
    if len(jadwal) == 0:
      print("Belum ada jadwal yang ditambahkan")
    else:
      for i in range(len(jadwal)):
        print(f"{i+1}. {jadwal[i]}")

  elif pilihan == 3:
    jadwal_tambahan = input("Masukkan jadwal: ")
    jadwal.insert(0, jadwal_tambahan)
    print("Jadwal berhasil ditambahkan")

  elif pilihan == 4:
    print("\n===== Hapus Jadwal =====")
    if len(jadwal) == 0:
      print("Tidak ada jadwal yang bisa dihapus")
    else:
      for i in range(len(jadwal)):
        print(f"{i+1}. {jadwal[i]}")
      hapus = int(input("Hapus nomor:"))
      jadwal.pop(hapus -1)
      print("Jadwal berhasil dihapus")

  elif pilihan == 5:
    print("\n===== 3 Jadwal Pertama =====")
    if len(jadwal) == 0:
      print("Tidak ada jadwal")
    else:
      for i in range(3):
        print(f"{i+1}. {jadwal[i]}")
  else:
    print("Selesai!")
    break
```
