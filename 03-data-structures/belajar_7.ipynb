```python
# Contoh
makanan = ("Seblak", "Ayam Goreng", "Cirambay")

# makanan.pop() # Ini akan menimbulkan error karena tuple bersifat immutable dan tidak memiliki method 'pop()'.

# Jika Anda ingin membuat tuple baru tanpa elemen terakhir, Anda bisa menggunakan slicing:
makanan_tanpa_terakhir = makanan[:-1]

print("Original tuple:", makanan)
print("Tuple without last element:", makanan_tanpa_terakhir)
# harus membuat tuple baru dulu, ga bisa ngubah tuple lama nanti error
```

# Latihan

```python
# Buat tuple yang berisi 5 nama teman.

# Tampilkan:
# Semua isi tuple.
# Teman pertama.
# Teman terakhir.
# Jumlah teman.

teman = ("Zidan", "Reza", "Aji", "Rafi", "Jaka")
print(f"Isi tuple ada {teman}")
print(f"Teman pertamaku adalah {teman[0]}")
print(f"Teman terakhirku adalah {teman[-1]}")
print(f"Jumlah temanku adalah {len(teman)}")
```

```python
# Tampilkan:
# Warna index ke-0.
# Warna index ke-2.
# Warna terakhir.

warna = ("pink", "biru", "putih", "ungu")

print(warna[0])
print(warna[2])
print(warna[3])
```

```python
# Tampilkan:
# Jumlah data menggunakan len().
# Total angka menggunakan sum().

angka = (4, 6, 15, 17, 9, 2022, 2026)

print(len(angka)) # Ada berapa data angka
print(sum(angka)) # Total penjumlahan data angka
```

```python
# Gunakan for untuk menampilkan:
# Hari ke-1 : Senin
# Hari ke-2 : Selasa
# Hari ke-3 : Rabu

hari = ("Senin", "Selasa", "Rabu")

for i in range(len(hari)):
  print(f"Hari ke-{i+1} : {hari[i]}")
```

```python
# Tampilkan:
# Apakah "Kucing" ada dalam tuple?
# Apakah "Ayam" ada dalam tuple?

hewan = ("Kucing", "Anjing", "Domba", "Serigala")

if "Kucing" in hewan:
  print(f"Kucing ada dalam tuple")

if "Ayam" in hewan:
  print(f"Ayam ada dalam tuple")
else:
  print(f"Ayam tidak ada dalam tuple")
# Fungsinya untuk mengecek Kucing dan Ayam maka if nya 2
# karena yang ditanyakan juga ada 2
```

```python
# Hitung berapa kali angka 2 muncul.

angka = (1, 1, 2, 2, 2, 2, 3, 3)

angka.count(2)

print(angka.count(2))
```

```python
# Mencari index
buah = ("Tomat", "Anggur", "Apel")

print(f"Tomat ada di index {buah.index("Tomat")}")
```

```python
# Tampilkan:
# 3 bulan pertama.
# 2 bulan terakhir.
# Februari sampai April.

bulan = (
    "Januari",
    "Februari",
    "Maret",
    "April",
    "Mei",
    "Juni",
    "Juli",
    "Agustus",
    "September",
    "Oktober",
    "November",
    "Desember",
)

print(bulan[:3])
print(bulan[-2:])
print(bulan[1:4])
```

# Projek

```python
# Kita akan menyimpan data film menggunakan tuple, karena data filmnya dianggap sebagai informasi tetap.

film = (
    "Interstellar",
    "Avengers",
    "Spirited Away",
    "Howl's Moving Castle"
)

while True:
  print("\n===== BIOSKOP CINDY =====")
  print("1. Lihat Daftar Film")
  print("2. Cari Film")
  print("3. Lihat Film Pertama")
  print("4. Lihat Film Terakhir")
  print("5. Jumlah Film")
  print("6. Keluar")

  pilihan = int(input("Masukkan pilihan Anda: "))

  if pilihan == 1:
    print("\n===== DAFTAR FILM =====")
    for i in range(len(film)):
      print(f"{i+1}. {film[i]}")

  elif pilihan == 2:
    nama_film = input("Masukkan nama film: ")
    if nama_film in film:
      print("Film tersedia!")
    else:
      print("Film tidak tersedia.")

  elif pilihan == 3:
    print(f"Film pertama adalah {film[0]}")

  elif pilihan == 4:
    print(f"Film terakhir adalah {film[-1]}")

  elif pilihan == 5:
    print(f"Jumlah film adalah {len(film)}")

  elif pilihan == 6:
    print("Anda keluar.")
    break

  else:
    print("Program tidak dapat membaca.")
```
