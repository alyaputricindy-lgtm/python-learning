# Mengenal Stuktur Data Dictionary

Dictionary adalah struktur data yang menyimpan data dalam bentuk:

key : value

```python
# Bentuk Dictionary
nama_dict = {
    "key1": "value",
    "key2": "value",
    "key3": "value"
}
```

```python
# Contoh sederhana

mahasiswa = {
    "nama": "Cindy",
    "umur": 17,
    "prodi" : "Sains Data Terapan"
}
```

```python
# Dictionary dapat memuat:

pak_tani = {
    "nama": "Petani Kode", # String
    "umur": 22, # Int
    "hobi": ["coding", "membaca", "cocok tanam"], # List
    "menikah": False, # Boolean
    "sosmed": { # Dictionary dalam dictionary
        "instagram": "@cinddissium",
        "facebook": "@cindy132"
    }
}
```

```python
# Mengambil data

mahasiswa = {
    "nama": "Cindy",
    "umur": 17
}

print(mahasiswa["nama"])
# Mengambil value berdasarkan key
```

```python
# Menambahkan data

mahasiswa = {
    "nama": "Cindy",
    "umur": 17
}

mahasiswa["kampus"] = "PENS"
print(mahasiswa)
```

```python
# Mengubah data

mahasiswa = {
    "nama": "Cindy",
    "umur": 17
}

mahasiswa["umur"] = 18
print(mahasiswa)
```

```python
# Menghapus data

mahasiswa = {
    "nama": "Cindy",
    "umur": 17
}

mahasiswa.pop("umur")
print(mahasiswa)
# kalau key nya diotak atik, value juga ikut kena otak atik
```

```python
# Mengecek key
# Apakah key itu ada di dalam dictionary

mahasiswa = {
    "nama" : "Cindy",
    "umur" : 17
}

print("nama" in mahasiswa)
# True karena key benar ada di dalam dictionary
```

```python
# len di Dictionary
# Ada berapa banyak pasangan key-value di Dictionary

mahasiswa = {
    "nama" : "Cindy",
    "umur" : 17
}

print(len(mahasiswa))
# Terdapat 2 pasangan key-value
```

```python
# Jika ingin mengambil key saja:

mahasiswa = {
    "nama" : "Cindy",
    "umur" : 17
}

for data in mahasiswa:
  print(data)
```

```python
# Jika ingin mengambil value saja:

mahasiswa = {
    "nama" : "Cindy",
    "umur" : 17
}

for data in mahasiswa:
  print(mahasiswa[data])
```

# Perbedaan List, Tuple, dan Dictionary

List: [] Mengambil index

Tuple: () Mengambil index

Dictionary: {} Mengambil key


```python
# List
buah = ["Apel", "Mangga", "Nangka"]
print(buah[1])

# Tuple
buah = ("Apel", "Mangga", "Nangka")
print(buah[1])

# Dictionary
buah = {
    "buah1": "Apel",
    "buah2": "Mangga",
    "buah3": "Nangka"
}
print(buah["buah2"])
```

# Latihan

```python
# Membuat dictionary
# Tampilkan:
# Nama
# Umur
# Jurusan

identitas = {
    "Nama": "Cindy",
    "Umur": 17,
    "Jurusan": "Sains Data Terapan"
}

print(identitas["Nama"])
print(identitas["Umur"])
print(identitas["Jurusan"])
```

```python
# Tampilkan semua data menggunakan for.

film = {
    "judul": "Interstellar",
    "tahun": 2014,
    "genre": "Sci-Fi"
}

for data in film:
  print(f"{data} : {film[data]}")
```

```python
# Lakukan:
# Tambahkan "stok": 10
# Ubah harga menjadi 7000000
# Tampilkan dictionary.

produk = {
    "nama": "Laptop",
    "harga": 7500000
}

produk["stok"] = 10
produk["harga"] = 7000000
print(produk)
```

```python
nilai = {
    "Matematika": 90,
    "Python": 95,
    "Statistika": 85
}

for data in nilai:
  print(f"{data} : {nilai[data]}")
```

```python
# Buat program yang menampilkan menu:
# ===== PROFIL =====
# 1. Lihat Profil
# 2. Tambah Data
# 3. Ubah Data
# 4. Hapus Data
# 5. Jumlah Data
# 6. Keluar

profil = {
    "nama": "Cindy",
    "umur": 19,
    "jurusan": "Sains Data",
    "kampus": "PENS"
}

while True:
  print("\n===== PROFIL =====")
  print("1. Lihat Profil")
  print("2. Tambah Data")
  print("3. Ubah Data")
  print("4. Hapus Data")
  print("5. Jumlah Data")
  print("6. Keluar")

  pilihan = int(input("Pilihan menu: "))

  if pilihan == 1:
    print("\n===== PROFIL ANDA =====")
    for data in profil:
      print(f"{data} : {profil[data]}")

  elif pilihan == 2:
    key = input("Masukkan data tambahan: ")
    value = input("Masukkan data tambahan: ")
    profil[key] = value
    print("Data berhasil ditambah!")

  elif pilihan == 3:
    key = input("Masukkan data yang ingin diubah: ")
    value = input("Masukkan data yang ingin diubah: ")
    profil[key] = value
    print("Data berhasil diubah!")

  elif pilihan == 4:
    key = input("Masukkan data yang ingin dihapus: ")
    profil.pop(key)
    print("Data berhasil dihapus!")

  elif pilihan == 5:
    print(f"Jumlah data: {len(profil)}")

  elif pilihan == 6:
    print("Terima kasih telah menggunakan program ini!")
    break

  else:
    print()
```
