## Operator Aritmatika

```python
a = int(input("Masukkan angka: "))
b = int(input("Masukkan angka:"))

# Penjumlahan (+)
c = a + b
print("Hasil penjumlahannya adalah", c)

# Pengurangan (-)
d = a - b
print("Hasil pengurangannya adalah", d)

# Perkalian (*)
e = a * b
print("Hasil perkaliannya adalah", e)

# Pembagian (/)
f = a / b
print("Hasil pembagiannya adalah", f)

# Sisa Bagi (%)
g = a % b
print("Hasil sisa baginya adalah", g)
"""
Misalkan 7 % 3
Maka 7 : 3 hasilnya 2 (sisa 1)
Lalu 3 x 2 hasilnya 6
Setelah itu 7 dikurangi 6 hasilnya 1
Sehingga hasil final 7 % 3 adalah 1
"""

# Pemangkatan (**)
h = a ** b
print("Hasil pemangkatannya adalah", h)
```

# Operator Bitwise

### **Bilangan Desimal (Angka Sehari-Hari)**

Ambil nilai dari
100 10 1

10² 10¹ 10⁰


Contoh: 345

 3 4 5

 3x100 + 4x10 + 5x1

 3x10² + 4x10¹ + 5x10⁰


### **Bilangan Biner (0 dan 1)**

Ambil nilai dari
8 4 2 1

Contoh: Ingin membuat angka 5 maka
8 4 2 1 sebagai bantuan

1.   Apakah 8 memenuhi angka 5?
Tidak karena 8 lebih besar dari 5 sehingga 0
2.   Apakah 4 memenuhi angka 5? Iya, maka 5-4 = 1 sehingga 0. Lalu sisanya 1 sebagai acuan baru bukan 5 lagi
3.   Apakah 2 memenuhi angka 1? Tidak karena 2 lebih besar dari 1 sehingga 0
4.   Apakah 1 memenuhi angka 1? Iya, maka 1-1 = 0 sehingga 1

Didapatkan bahwa 5 = 0101

Gampangnya juga bisa menjadi seperti berikut:
0x8 + 1x4 + 0x2 + 1x1
= 4 + 1
= 5





```python
# AND (&)
a = 5
b = 3
print(a & b)
# 5 adalah 0101
# 3 adalah 0011
# -------- 0001
# Outputnya 1

c = 6
d = 2
print(c & d)
# c adalah 0110
# d adalah 0010
# -------- 0010
# Outputnya adalah 2

e = 7
f = 4
print (e & f)
# f adalah 0111
# d adalah 0100
# -------- 0100
# Outputnya adalah 4

g = 8
h = 1
print(g & h)
# g adalah 1000
# h adalah 0001
# -------- 0000
# Outputnya adalah 0


# OR (|)
angka1 = 2
angka2 = 7
print(angka1 | angka2)
# 2 adalah 0010
# 7 adalah 0111
# -------- 0111
# Outputnya 7 karena hasil (0111) selaras dengan 7 (0111)

# XOR (^)
# Kalau beda hasilnya 1, kalau sama hasilnya 0
angka = 5
angkaaa = 3
print(angka ^ angkaaa )
# 5 adalah 0101
# 3 adalah 0011
# -------- 0110
# Lalu ubah ke desimal
# 8 4 2 1
# 0 1 1 0
# Maka 4 + 2 = 6 sehingga outputnya 6

# Negasi/Kebalikan (~)
y = 5
print(~y)
# Anggap aja gini deh:
# -(5+1) = -6
# Intinya minus lalu ditambah 1

# Left Shift (<<)

# Right Shift
```
