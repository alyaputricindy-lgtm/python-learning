## Penulisan Statement di Python

Statement adalah sebuah instruksi atau kalimat perintah dari kita buat Python biar Python bisa bekerja.

```python
# 1 statement
# Tidak diakhiri titik dan koma
print("Hello, world!")
```

```python
# 4 statement
print("Hello, world!") # Statement ke 1
print("I'm ready to going to PENS") # Statement ke 2
print("But not ready for Pemrograman") # Statement ke 3
nama = "Cindy" # Statement ke 4
```

```python
# Kalau beberapa statement ingin digabung jadi 1 baris menggunakan tanda ;
# Tidak dianjurkan karena aku jadi bingung jir

print("Hello"); print("World"); print("Tutorial Python untuk Pemula")
nama_depan = "petani"; nama_belakang = "kode"
```

## Penulisan Case

Bagi Python, kata "Makan" dan "makan" itu berbeda makna. Jadi harus memperhatikan huruf kecil dan huruf besarnya ya

```python
# Snake Case digunakan pada:
# Memisahkan dua kata pakai underscore
module_name, package_name, method_name, function_name, , global_var_name, instance_var_name, function_parameter_name, local_var_name.

# CamelCase digunakan Pada:
# Memisahkan dua kata dengan cara digabung dan tiap huruf pertama dikapitalkan
ClassName, ExceptionName

# ALL CAPS digunakan Pada:
# Kalau kapital semua pakai underscore
GLOBAL_CONSTANT_NAME
```
